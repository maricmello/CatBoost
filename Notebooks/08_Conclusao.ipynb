## Aprendizados

1. **Vazamento também existe dentro do algoritmo.** A ideia do CatBoost é a mesma regra da separação treino e teste, e o E5 mostrou o custo de ignorá-la
2. **No boosting, mais árvores podem piorar o modelo.** Early stopping e árvores rasas fizeram diferença
3. **Acurácia engana com classes desbalanceadas.** ROC-AUC, PR-AUC, calibração e um limiar escolhido na validação mostram muito mais
4. **Algoritmo mais sofisticado não garante ganho grande.** Desempenho, tempo e interpretação entram na escolha
5. **As variáveis limitam o modelo.** Parte dos churns não é previsível com o que a base tem

## Quando usar o CatBoost

| Usaria | Pensaria duas vezes |
|:--|:--|
| Muitas categóricas, principalmente com muitos valores (loja, produto, CEP) | Bases muito pequenas |
| Pouco tempo para ajustar parâmetros | Quando a explicação precisa ser por coeficientes |
| Predição precisa ser rápida | Quando um modelo linear já chega ao mesmo resultado |
